#### What is Simple Linear Regression

Simple Linear Regression is a statistical method used to model the relationship between a dependent variable (Y) and a single independent variable (X) using a straight-line equation

#### What are the key assumptions of Simple Linear Regression

For Simple Linear Regression to provide reliable results, the following assumptions must be met:

1.Linearity 2.Independence 3.Homoscedasticity 4.Normality of Residuals 5.No Multicollinearity 

#### What does the coefficient m represent in the equation Y=mX+c

The coefficient m represents the slope of the line. It indicates how much Y (the dependent variable) changes for a one-unit increase in X (the independent variable).

#### What does the intercept c represent in the equation Y=mX+c

The intercept c (also called the y-intercept) represents the value of Y when X=0. In other words, it's the point where the line crosses the Y-axis.

#### What is the purpose of the least squares method in Simple Linear Regression

The Least Squares Method is used in Simple Linear Regression to find the best-fitting straight line that minimizes the error between the actual data points and the predicted values.

#### How is the coefficient of determination (R²) interpreted in Simple Linear Regression

The coefficient of determination (𝑅*R) is a statistical measure that tells us how well the regression model explains the variability in the dependent variable Y based on the independent variable X.


#### What is Multiple Linear Regression

Multiple Linear Regression (MLR) is an extension of Simple Linear Regression, where the dependent variable (Y) is predicted using two or more independent variables (𝑋1,X2,X3,...)


#### What is the main difference between Simple and Multiple Linear Regression

 Simple Linear Regression is used when only one independent variable affects Y. Multiple Linear Regression is used when multiple factors influence Y, leading to a more accurate prediction

#### What are the key assumptions of Multiple Linear Regression

To ensure that a Multiple Linear Regression (MLR) model provides reliable and accurate predictions, the following assumptions must be met:
1. No Multicollinearity  2.Linearity  3.Homoscedasticity  4. Independence of Errors  5. Normality of Residuals 6.No Omitted Variable Bias

#### What is heteroscedasticity, and how does it affect the results of a Multiple Linear Regression model

Heteroscedasticity occurs when the variance of the residuals (errors) is not constant across all levels of the independent variables in a regression model.

In an ideal Multiple Linear Regression (MLR) model, residuals (the differences between actual and predicted values) should be randomly distributed with constant variance (homoscedasticity).

#### How can you improve a Multiple Linear Regression model with high multicollinearity

Multicollinearity occurs when two or more independent variables in a regression model are highly correlated. This can lead to unstable coefficient estimates, making it difficult to determine the individual effect of each variable on the dependent variable.

#### What is the role of interaction terms in Multiple Linear Regression

Interaction terms in Multiple Linear Regression capture the combined effect of two or more independent variables on the dependent variable. They help when the effect of one predictor depends on the value of another predictor.

#### Why is it important to scale variables in Multiple Linear Regression

Scaling means standardizing or normalizing independent variables (𝑋) so they have a consistent range and unit. This is especially important when variables are measured on different scales (e.g., income in dollars vs. age in years).

#### What is polynomial regression

Polynomial Regression is a type of linear regression where the relationship between the independent variable (X) and the dependent variable Y) is modeled as an n-th degree polynomial.

####  How does polynomial regression differ from linear regression

Both Polynomial Regression and Linear Regression are types of regression models, but they differ in how they model relationships between variables.

#### When is polynomial regression used

Plynomial regression is used when the relationship between the independent variable(s) and the dependent variable is non-linear but can be modeled using a polynomial function. It allows for a curved fit instead of a straight-line relationship seen in simple linear regression.

#### What is the general equation for polynomial regression

Polynomial regression is an extension of linear regression where the relationship between the independent variable X and the dependent variable Y is modeled as an n-th degree polynomial.

####  Can polynomial regression be applied to multiple variables

Polynomial regression can be extended to multiple variables (also known as Multiple Polynomial Regression). It allows us to model complex relationships between two or more independent variables and a dependent variable.
